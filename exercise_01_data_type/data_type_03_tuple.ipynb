# 튜플 자료형 학습과 연습

```python
t1 = ()
print(t1)
t2 = (1,)
print(t2)
t3 = (1, 2, 3)
print(t3)
t4 = 1, 2, 3
print(t4)
t5 = ('a', 'b', ('ac', 'ad'))
print(t5)
```

* 튜플의 요솟값을 지우거나 변경하려고 하면 어떻게 될까?

### 1. 튜플 요솟값을 삭제하려 할 때

```python
t1 = (1, 2, 3)
del t1[0]
```

### 2. 튜플 요솟값을 변경하려 할 때

```python
t1 = (1, 2, 3)
t1[2] = 4
```

##  튜플 다루기

### 1. 인덱싱하기

```python
t1 = (1, 2, 'a', 'b', ('ac', 'ad'))
print(t1[-1])
print(t1[0])
print(t1[3])
print(t1[4])
```

### 2. 슬라이싱하기

```python
t1 = (1, 2, 'a', 'b', ('ac', 'ad'))
print(t1[1:])
print(t1[:])
print(t1[:-1])
```

### 3. 튜플 더하기

```python
t1 = (1, 2, 3)
t2 = (4, 5, 6)
print(t1 + t2)
```

### 4. 튜플 곱하기

```python
t1 = (3, 5)
print(t1 * 3)
```

### 5. 튜플 길이 구하기

```python
t1 = (1, 2, 3, 'a', 'b', ('ac', 'ad'))
print(len(t1))
```

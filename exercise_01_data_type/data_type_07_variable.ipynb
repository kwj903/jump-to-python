## 변수란?

```python
# 변수는 메모리에 주소 값을 가지고 있다.
a = [1, 2, 3]
print(id(a))
```

## 리스트를 복사할 때

```python
a = [1, 2, 3]
b = a

print(id(a))
print(id(b))
print(a is b)

a[1] = 4
print(a)
print(b)
print(a is b)
```

### 1. [:]사용

```python
a = [1, 2, 3]
b = a[:]
print(a is b)

a[1] = 4
print(a)
print(b)
```

### 2. copy 모듈 사용

```python
from copy import copy
a = [1, 2, 3]
b = copy(a)

print(a)
print(b)
print(b is a)
print(b == a)
print(id(a))
print(id(b))
```

## 변수를 만드는 여러 가지 방법

```python
a, b = ('python', 'life')
print(a)
print(isinstance(a, tuple))
print(b)
print(isinstance(b, tuple))

(a, b) = 'python', 'life'
print(a)
print(isinstance(a, str))
print(b)
print(isinstance(b, str))

a, b = 'python', 'life'
print(a)
print(isinstance(a, str))
print(b)
print(isinstance(b, str))

[a, b] = ['python', 'life']
print(a)
print(isinstance(a, list))
print(b)
print(isinstance(b, str))

a = b = 'python'
print(a == b)
print(a)
print(b)

a = 3
b = 5
a, b = b, a
print(a)
print(b)
```

- 나 혼자 코딩문제

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)

# False가 뜨는 이유는 각 변수는 다른 메모리 주소값을 할당받기 때문이다.
print(a == b)
# True가 뜨는 이유는 a와 b의 값이 같기 때문이다.
```

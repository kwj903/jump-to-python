# 딕셔너리 자료형 학습과 연습

## 딕셔너리란?
key, value 한 쌍으로 구성되어 있는 자료형

## 딕셔너리는 어떻게 만들까?

```python
dic = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(dic)
print(dic['name'])
print(dic["phone"])
print(dic["birth"])

a = {1:'a'}
print(a)
a = {'a':[1, 2, 3]}
print(a)
```

## 딕셔너리 쌍 추가, 삭제하기

### 1. 딕셔너리 쌍 추가하기

```python
a = {1:'a', 2:'b'}
a[3] = 'c'
print(a)
print(a[3])

a['name'] = 'woojae'
print(a)
print(a['name'])

a[4] = [1, 2, 3]
print(a)
print(a[4])


```

### 2. 딕셔너리 요소 삭제하기

```python
a = {1: "a", 2: "b", 3: "c", "name": "woojae", 4: [1, 2, 3]}
print(a)
del a[4]
print(a)
```

### 딕셔너리에서 Key 사용해 Value 얻기

```python
grade = {'seulgi':100, 'woojae':90, 'yeri':80}
print(grade['seulgi'])
print(grade['yeri'])
print(grade['woojae'])

a = {1: "a", 2: "b", 3: "c", "name": "woojae", 4: [1, 2, 3]}
print(a[1])
print(a[3])

a = {'a':1, 'b':2, 'c':3}
print(a['c'])
print(a['a'])
print(a['b'])

dic = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(dic['name'])
print(dic['phone'])
print(dic['birth'])
```

### 딕셔너리를 만들 때 주의할 사항

```python
# 중복되는 키값을 설정하면 하나를 제외하고 나머지는 무시됨
a = {1:'a', 1:'b', 1:'c'}
print(a)

# Key에 리스트를 쓸 수 없다. Key에는 변하지 않는 값을 써야한다.
a = {[1, 2, 3]:'a'}
print(a)
```

## 딕셔너리 관련 함수

### Key 리스트 만들기(keys)

```python
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(a.keys())
```

* 파이썬 3.0 이후 버전의 keys 함수, 어떻게 달라졌나?
 a.key() 함수를 호출하면 반환값으로 이전에는 dict_keys가 아닌 리스트를 반환했는데 3.0이후로는 메모리 낭비를 위해 dict_keys 객체를 반환한다. 이러한 객체는 기본적인 반복(iterate)구문을 실행 할 수 있다.

```python
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
for k in a.keys():
    print(k)
    
b = list(a.keys())
print(b)
```

### Value 리스트 만들기(values)

```python
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(a.values())

b = list(a.values())
print(b)

c = [a['name'], a['birth']]
print(c) 
```

### Key, Value 쌍 얻기(items)

```python
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(a.items())
```

### Key:Value 쌍 모두 지우기(clear)

```python
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(a)
a.clear()
print(a)
```

### Key로 Value 얻기(get)

```python
# get은 없는 키값을 입력해도 에러가 나지 않고 None이 출력됨
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print(a)
b = a.get('name')
print(b)

c = a.get('monkey')
print(c)
# d = a['monkey']
# print(d)

# 키가 없을 경우 디폴드 값을 정해서 가져오게하는 방법도 있다.
e = a.get('monkey', '없음')
print(e)
```

### 해당 Key가 딕셔너리 안에 있는지 조사하기(in)

```python
a = {'name':'woojae', 'phone':'01091620623', 'birth':'910909'}
print('name' in a)
print('name' not in a)
print('monkey' in a)
print('monkey' not in a)
```

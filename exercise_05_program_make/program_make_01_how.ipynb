### 업그레이드한 나만의 구구단 프로그램
- 프로그램 목표 : 사용자가 알고싶은 숫자의 구구단을 알기 쉽게 출력해주고 한번 실행하면 계속 또 다시 다른 단수를 출력 할 수 있고 끝내는 것도 사용자의 선택으로 끝낼 수 있게끔 한다.
- 입력값 : 알고자하는 숫자의 구구단에 대한 수
- 입력방식 : 사용자가 알고 싶은 단수를 입력한다. 원하는 출력값이 나오면 또 다시 알고싶은 단수를 입력 할 수 있어야하며, 구구단 프로그램을 마치는 입력도 있어야한다.
- 출력값 : 예를 들어 2 * 1 = 2 라는 형태로 알아 보기 쉬운 형태로 출력한다.
- 결과값을 담는 방식 : 리스트로 결과값을 담고 리스트에 담긴 해당 숫자의 곱셈 값을 출력에 함께 넣어서 원하는 출력값이 나오도록한다.
- 다듬기 : input으로 받은 값은 문자열값으로 취급되기 때문에 숫자연산에 이용하려면 무조건 int()내장함수를 써줘서 숫자타입으로 바꿔줘야한다. 끝내는 입력값을 받을 때는 바로 끝낼 수 있도록  
조건문을 반복문 상단에 배치하고 간단하게 break문으로 프로그램을 끝낸다.

```python
def gugu():
    while True:
        num = input('알고싶은 구구단 숫자를 입력하시오(끝:/): ')
        result = []
        if num == '/':
            print('-'*20)   
            print('프로그램을 종료합니다.')
            break
        num = int(num)
        print('-'*20)
        print(f'{num}단!!!!')
        for i in range(1, 10):
            result.append(num * i)
            print(f'{num} * {i} = {result[0]*i}')
            
gugu()
```

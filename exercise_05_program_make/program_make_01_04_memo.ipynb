1. 우선 다음과 같이 입력으로 받은 옵션과 메모를 출력하는 코드를 작성해 보자.

```python
# /Users/kwj903/Downloads/workspace/education/my_educate/jump-to-python/exercise_05_program_make/memo.py
import sys

option = sys.argv[1]
memo = sys.argv[2]

print(option)
print(memo)
# sys.argv는 프로그램을 실행할 때 입력된 값을 읽어 들일 수 있는 파이썬 라이브러리이다.
# sys.arvg[0]는 입력받은 값 중에서 파이썬 프로그램 이름인 memo.py이브로 우리가 만들려는 기능에는 필요 없는 값이다.
# sys.arvg[1]은 옵션값이 되고, sys.arvg[2]는 메모 내용이 된다.
```

3. 이제 입력으로 받은 메모를 파일에 쓰도록 코드를 변경해 보자.

```python
# /Users/kwj903/Downloads/workspace/education/my_educate/jump-to-python/exercise_05_program_make/memo.py
import sys

option = sys.argv[1]

if option == '-a':
    memo = sys.argv[2]
    f = open("memo.txt", 'a')
    f.write(memo)
    f.write('\n')
    f.close()
```

5. 이번에는 작성한 메모를 출력하는 부분을 만들 차례이다. 메모 출력은 다음과 같이 동작하도록 만들어 보자.

`python memo.py -v`

```python
# /Users/kwj903/Downloads/workspace/education/my_educate/jump-to-python/exercise_05_program_make/memo.py
import sys

option = sys.argv[1]

if option == "-a":
    memo = sys.argv[2]
    f = open("memo.txt", "a")
    f.write(memo)
    f.write("\n")
    f.close()

if option == '-v':
    f = open('memo.txt', 'r')
    memo = f.read()
    f.close()
    print(memo)
```

6. 코드를 수정한 후 다음과 같은 명령을 수행해 보자.

- `python memo.py -v`

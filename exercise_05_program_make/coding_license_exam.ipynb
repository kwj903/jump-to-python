## Q5 피보나치 함수  
첫 번째 항의 값이 0이고 두 번째 함의 값이 1일 때, 이후에 이어지는 항은 이전의 두 항을 더한 값으로 이루어지는 수열을 피보나치 수열이라고 한다.  
`0, 1, 1, 2, 3, 5, 8, 13 ...`  
입력을 정수 n으로 받았을 때, n 이하까지의 피보나치 수열을 출력하는 함수를 작성해 보자.

````python
def fibo(num):
    fibonum = [0, 1]
    while True:
        if fibonum[-1] < num:
            fibonum.append(fibonum[-1] + fibonum[-2])
        elif fibonum[-1] == num:
            return fibonum
        else:
            fibonum.pop()
            return fibonum
            
print(fibo(345))
        
````

## Q10 사칙연산 계산기  
다음과 같이 동작하는 클래스 Calculator를 작성하시오.  

````python
cal1 =Calculator([1,2,3,4,5])
cal1.sum()
cal1.avg()
cal2 = Calculator([6,7,8,9,10])
cal2.sum()
cal2.avg()
````

````python
class Calculator:
    def __init__(self, numbers):
        self.numbers = numbers

    def sum(self):
        return sum(self.numbers)

    def avg(self):
        return sum(self.numbers) / len(self.numbers)
````

````python
cal1 = Calculator([1, 2, 3, 4, 5])
print(cal1.sum())
print(cal1.avg())
cal2 = Calculator([6, 7, 8, 9, 10])
print(cal2.sum())
print(cal2.avg())
````

## Q13 DashInsert 함수  
DashInsert 함수는 숫자로 구성된 문자열을 입력받은 뒤 문자열 안에서 홀수가 연속되면 두 수 사이에 - 를 추가하고, 짝수가 연속되면 * 를 추가하는 기능을 갖고 있다. DashInsert 함수를 완성하시오.  
입력 예시: 4546793  
출력 예시: 454*67-9-3

````python
numbers = "4546793"

def DashInsert(numbers):
    numbers = list(numbers)
    result = []
    for i in range(0, len(numbers)):
        result.append(numbers[i])
        if i != len(numbers)-1:
            if int(numbers[i]) % 2 == 0 and int(numbers[i+1]) % 2 == 0:
                result.append("*")
            elif int(numbers[i]) % 2 == 1 and int(numbers[i+1]) % 2 == 1:
                result.append("-")
    result = "".join(result)
    return result
print(DashInsert(numbers))
````

## Q14 문자열 압축하기  
문자열을 입력받아 같은 문자가 연속적으로 반복되는 경우에 그 반복 횟수를 표시해 문자열을 압축하여 표시하시오.  
입력 예시: aaabbcccccca  
출력 예시: a3b2c6a1  

````python
sentence = "aaabbcccccca"

def str_zip(strnum):
    strnum = list(strnum)
    result = []
    count = 1
    for i in range(1, len(strnum)):
        if strnum[i] == strnum[i - 1]:
            count += 1
        else:
            result.append(strnum[i - 1])
            result.append(str(count))
            count = 1
    result.append(strnum[-1])
    result.append(str(count))
    result = "".join(result)
    return result

print(str_zip(sentence))
````

## Q16 모스 부호 해독  
문자열 형식으로 입력받은 모스 부호(dot: . dash:-)를 해독하여 영어 문장으로 출력하는 프로그램을 작성하시오.  
* 글자와 글자 사이는 공백 1개, 단어와 단어 사이는 공백 2개로 구분한다.  
* 예를 들어 다음 모스 부호는 "HE SLEEPS EARLY"로 해석해야 한다.  
`.... .  ... .-.. . . .--. ...  . .- .-. .-.. -.--`

````python
import string




def morse_str(mos):
    letters = list(string.ascii_uppercase)
    morse = [
        ".-",
        "-...",
        "-.-.",
        "-..",
        ".",
        "..-.",
        "--.",
        "....",
        "..",
        ".---",
        "-.-",
        ".-..",
        "--",
        "-.",
        "---",
        ".--.",
        "--.-",
        ".-.",
        "...",
        "-",
        "..-",
        "...-",
        ".--",
        "-..-",
        "-.--",
        "--..",
    ]
    morse_dict = dict(zip(morse, letters))
    mos = mos.split(" ")
    result = []
    for i in mos:
        result.append(morse_dict.get(i, " "))
    result = "".join(result)
    return result
morse_sentence = ".... .  ... .-.. . . .--. ...  . .- .-. .-.. -.--"
print(morse_str(morse_sentence))
````

## Q17 기초 메타 문자
다음 중 정규식 `a[.]{3,}b`과 매치되는 문자열은 무엇일까?  
1. acccb  
2. a....b  
3. aaab  
4. a.cccb  

````python
import re
num_list = ["acccb", "a....b", "aaab", "a.cccb"]
print(num_list)

def pic_num(num_list):
    for i, sentence in enumerate(num_list):
        p = re.compile("a[.]{3,}b")
        m = p.match(sentence)
        num = i+1
        if m:
            return "정답: " + str(num) + "번"

print(pic_num(num_list))

# 답은 2번
````

## Q20 전방 탐색  
다음은 이메일 주소를 나타내는 정규식이다. 이 정규식은 park@naver.com, kim@daum.net, lee@myhome.co.kr 등과 매치된다. 긍정형 전방 탐색 기법을 사용하여 .com, .net이 아닌  
이메일 주소는 제외시키는 정규식을 작성하시오.  
`.*[@].*[.].*$`

````python
import re

juso = """
park@naver.com
kim@daum.net
lee@myhome.co.kr
"""
p = re.compile(".*[@].*[.](?!com|net).*$")
m = p.search(juso)
print(m.group())
````

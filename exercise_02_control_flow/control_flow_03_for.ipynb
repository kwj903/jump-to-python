# for문 학습과 연습

## for 문의 구조

### 예제를 통해 for문 이해하기

#### 1. 전형적인 for문

```python
test_list = ["one", "two", "three"]
for i in test_list:
    print(i)
```

#### 2. 다양한 for문의 사용

```python
a = [(1, 2), (3, 4), (5, 6)]
for (first, last) in a:
    print(first + last)
```

#### 3. for문의 응용

```python
marks = [90, 25, 67, 45, 80]
number = 0
for mark in marks:
    number = number + 1
    if mark >= 60:
        print("%d번 학생은 합격입니다." % number)
    else:
        print("%d번 학생은 불합격입니다." % number)
```

## for문과 continue문

```python
marks = [90, 25, 67, 45, 80]
number = 0
for mark in marks:
    number += 1
    if mark < 60: continue
    print("%d번 학생 축하합니다. 합격입니다." % number)
```

## for문과 함께 자주 사용하는 range 함수

```python
a = range(10)
print(a)
print(list(a))
```

```python
a = range(1, 11)
print(a)
print(list(a))
```

### range 함수의 예시 살펴보기

```python
add = 0
for i in range(1, 11):
    add += i
print(add)
```

```python
marks = [90, 25, 67, 45, 80]
for number in range(len(marks)):
    if marks[number] < 60: continue
    print("%d번 학생 축하합니다. 합격입니다." % (number + 1))
```

- ***나 혼자 코딩*** : for문과 range 함수를 사용하여 1부터 100까지 더해 보자

```python
num = 0
for i in range(1, 101):
    num += i
print(num)
```

### for와 range를 이용한 구구단

```python
for i in range(2, 10):
    print("구구단 %d단" % i)
    for j in range(1, 10):
        print("%d * %d = %d" % (i, j, i*j))
    print("_" * 20)
print("이로써 구구단을 완료했습니다.", end="    ~~~~끝~~~~   ")
```

* ***점프 투 파이썬*** : 매개변수 end를 넣어 준 이유는 무엇을까?

##### 매개변수 end는 print문이 끝난 후 다음 줄로 넘어가는 것을 방지하고, 대신 end에 지정된 문자열을 출력한다. end의 기본값은 "\n"이다.  
##### print문에는 end 매개변수 외에도 sep 매개변수가 있다. sep는 print문에서 여러 값을 출력할 때 값 사이에 출력되는 문자열을 지정한다. sep의 기본값은 " "이다. 예를 들어, print("Python", "Java", sep=", ")라고 하면 "Python, Java"가 출력된다.  
##### 아무것도 출력하지 않고 다음 줄로 넘어가지 않으려면 end 매개변수에 빈 문자열 ""을 지정하면 된다. 예를 들어, print("Hello", end="")라고 하면 "Hello"가 출력되고 다음 줄로 넘어가지 않는다.  
##### print문에서 end 매개변수와 sep 매개변수를 적절히 활용하면 출력 형식을 다양하게 조절할 수 있다. 예를 들어, print("Python", "Java", sep=", ", end="!\n")라고 하면 "Python, Java!"가 출력되고 다음 줄로 넘어간다.  
##### print문에서 end 매개변수를 입력하지 않으면 기본값인 "\n"이 사용되어 출력 후 다음 줄로 넘어간다. 따라서 print("Hello")라고 하면 "Hello"가 출력되고 다음 줄로 넘어간다.  

## 리스트 내포 사용하기

```python
a = [1, 2, 3, 4]
result = []
for i in a:
    result.append(i * 3)
print(result)
```

```python
a= [1,2,3,4]
result = [i *3 for i in a]
print(result)
```

```python
a = [1,2,3,4]
result = [i *3 for i in a if i  % 2 ==0]
print(result)
```

```python
result = [i*j for i in range(2, 10) for j in range(1, 10)]
print(result)
```

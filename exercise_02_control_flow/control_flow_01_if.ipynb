## if문은 왜 필요할까?

```python
money = True
if money:
    print("택시를 타고 가라")
else:
    print("걸어가라")
```

### and, or, not

```python
money = 2000
card = True
if money >= 3000 or card:
    print("택시를 타고 가라")
else:
    print("걸어가라")
```

```python
# 만약 주머니에 돈이 있으면 택시를 타고 가라. 없으면 걸어가라.
poket = ['paper', 'cellphone', 'money']
if 'money' in poket:
    print("택시를 타고 가라")
else:
    print("걸어가라")
```

- 나 혼자 코딩!

```python
# '주머니에 카드가 없다면 걸어가고, 있다면 버스를 타고 가라'라는 문장을 조건문으로 만들어보자.
poket = ['paper', 'cellphone', 'money']
if 'card' not in poket:
    print("걸어가라")
else:
    print("버스를 타고 가라")
```

## 다양한 조건을 판단하는 elif

```python
# 주머니에 돈이 있으면 택시를 타고, 주머니에 돈은 없지만 카드가 있으면 택시를 타고, 돈도 없고 카드도 없으면 걸어 가라.
poket = ['paper', 'cellphone']
if 'money' in poket:
    print("택시를 타고 가라")
else:
    if 'card' in poket:
        print("택시를 타고 가라")
    else:
        print("걸어가라")
```

```python
# elif를 사용하면 위의 조건문을 좀더 간결하게 만들 수 있다.
poket = ['paper', 'cellphone']
if 'money' in poket:
    print("택시를 타고 가라")
elif 'card' in poket:
    print("택시를 타고 가라")
else:
    print("걸어가라")
```

## 조건부 표현식

```python
score = 70

if score >= 60:
    message = "success"
    print(message)
else:
    message = "failure"
    print(message)
# 아래 조건부 표현식을 해석하는 방법 : message는 "success" 이다. 단, score가 60보다 크거나 같다면, 하지만 score가 60보다 작다면 message는 "failure" 이다.    
message = "success" if score >= 60 else "failure"
print(message)
```

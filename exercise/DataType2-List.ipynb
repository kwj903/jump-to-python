# 리스트 자료형 공부와 연습

## 리스트는 어떻게 만들고 사용할까?

```python
# 리스트의 표현
odd = [1, 3, 5, 7, 9]
print(odd)
# 리스트의 여러가지 생김새
a = []
print(a)
b = [1, 3, 5, 7, 9]
print(b)
c = ['Life', 'is', 'too', 'short']
print(c)
d = [1, 3, "too", "short"]
print(d)
e = [1, 3, ['Life', 'is']]
print(e[2])
```

## 리스트 인덱싱과 슬라이싱

### 리스트의 인덱싱

```python
a = [1, 2, 3]
print(a)
print(a[0]) 
print(a[0]+a[2])
print(a[-1])

a = [1, 2, 3, ['a', 'b', 'c']]
print(a[0])
print(a[-1])
print(a[3])
print(a[-1][0])
print(a[-1][1])
print(a[-1][2])
print(a[-1][-1])
```

- 삼중 리스트에서 인덱싱하기

```python
a = [1, 2, ['a', 'b', ['Life', 'is']]]
print(a[2][2][0])
```

### 리스트의 슬라이싱

```python
a = [1, 2, 3, 4, 5]
print(a[0:2])
b = "12345"
print(b[0:2])

aa = a[:2]
print(aa)
ab = a[2:]
print(ab)
```

#### 나혼자 코딩 문제

```python
A =[1, 2, 3, 4, 5]
a = A[1:3]
print(a)
```

- 중첩된 리스트에서 슬라이싱하기

```python
a = [1, 2, 3, ['a', 'b', 'c'], 4, 5]
print(a[2:5])
print(a[3][:2])
```

### 리스트 연산하기

```python
# 1리스트 더하기(+)
a = [1, 2, 3]
b = [4, 5, 6]
print(a + b)

# 리스트 반복하기(*)
a = [1, 2, 3]
print(a * 3)

# 리스트 길이 구하기(len)
a = [1, 2, 3]
print(len(a))
```

- 초보자가 실수하기 쉬운 리스트 연산 오류

```python
a = [1, 2, 3]
# print(a[2] + "hi")
print(str(a[2]) + "hi")
```

### 리스트의 수정과 삭제

```python
# 리스트에서 값 수정하기
a = [1, 2, 3]
print(a)
a[2] = 4
print(a)

# del 함수 사용하기
a = [1, 2, 3]
print(a)  
del a[2]
print(a) 
a = [1, 2, 3, 4, 5]
print(a)
del a[1:]
print(a)
```

### 리스트 관련 함수

```python
# 리스트에 요소 추가하기(append)
a = [1, 2, 3]
a.append(4)
print(a)
a.append([5, 6])
print(a)

# 리스트 정렬(sort)
a = [1, 4, 3, 2]
print(a)
a.sort()
print(a)

a = ['a', 'c', 'b']
print(a)
a.sort()
print(a)

# 리스트 뒤집기(reverse)
a = ['a', 'c', 'b']
print(a)
a.reverse()
print(a)

# 위치 반환(index)
a = [1, 2, 3]
print(a.index(1))
print(a.index(2))
print(a.index(3))
# print(a.index(0))

# 리스트에 요소 삽입하기(insert)
a = [1, 2, 3]
a.insert(1, 55)
print(a)
a.insert(3, 66)
print(a)

# 리스트 요소 제거하기(remove)
a = [1, 2, 3, 1, 2, 3]
a.remove(1)
print(a)
a.remove(1)
print(a)

# 리스트 요소 끄집어내기(pop)
a = [1, 2, 3]
print(a)
print(a.pop())
print(a)
a = [1, 2, 3]
print(a.pop(0))
print(a)

# 리스트에 포함된 요소 x의 개수 세기(count)
a = [1, 2, 3, 1]
print(a.count(1))

# 리스트 확장하기(extend)
a = [1, 2, 3]
print(a)
a.extend([6,7])
print(a)
b = [4, 5]
a.extend(b)
print(a)




```

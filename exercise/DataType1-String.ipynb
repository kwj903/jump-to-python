## 인덱싱과 슬라이싱 연습문제

```python
# 문자열 슬라이싱
a = "20010331Rainy"
print(a[:])
year = a[:4]
print(year)
month = a[4:6]
print(month)
day = a[6:8]
print(day)
whether = a[8:]
print(whether)
print(year + "년 " + month + "월 " + day + "일 " + whether + "입니다.")
```

```python
# 문자열 치환
a = 'pithon'
print(a)
b = a[1]
print(b)
b = 'y'
print(b)
newString = a[:1] + b + a[2:]
print(newString)
newString2 = a.replace('i', 'y')
print(newString2)
newString3 = list(a)
newString3[1] = b
print(type(newString3))
newString3 = ''.join(newString3)
print(type(newString3))
print(newString3)
```

## 문자열 포메팅 따라하기

```python
# 숫자 바로 대입
a = "I eat %d apples."
print(a)
print(a %3)

# 문자열 바로 대입
b = "I eat %s apples."
print(b)
print(b % "five")

# 숫자 값을 나타내는 변수로 대입
number = 3
c = a % number
print(c)

# 문자열, 숫자 값을 나타내는 변수로 대입
num = 10
day = "three"
d = "I eat %d apples. so I was sick for %s days."
print(d)
print(d % (num, day))

# 정령과 공백
e = "Error is %d%%." % 98
print(e)
# e = "Error is %d%." % 98
# print(e)

print("%10s" % "hi")
print("%-10swoojae" % "hi")

# 소수점 표현하기
f = "%0.4f" % 3.42134234
print(f)
g = "%10.4f" % 3.42134234
print(g)
```

## format 함수를 사용한 포매팅

```python
# 숫자 바로 대입하기
a = "I eat {0} apples".format(3)
print(a)

# 문자열 바로 대입하기
b = "I eat {0} apples".format("five")
print(b)

# 숫자 값을 나타내는 변수로 대입하기
number = 3
c = "I eat {0} apples".format(number)
print(c)

# 2개 이상의 값 대입하기
num = 10
day = "three"
d = "I eat {0} apples. so I was sick for {1} days.".format(num, day)
print(d)

# 이름으로 대입하기
e = "I eat {number} apples. so I was sick for {day} days.".format(number=45, day=number)
print(e)

# 인덱스와 이름 혼용해서 대입하기
f = "I eat {0} apples. so I was sick for {day} days.".format(32, day="five")
print(f)

# 왼쪽 정렬
g = "{0:<10}".format("hi")
print(g)

# 오른쪽 정렬
h = "{0:>10}".format("hi")
print(h)

# 가운데 정렬
i = "{0:^10}".format("hi")
print(i)

# 공백 채우기
j = "{0:=^10}".format("hi")
print(j)
k = "{0:*^10}".format("hi")
print(k)

# 소수점 표현하기
y = 3.42134234
l = "{0:0.4f}".format(y)
print(l)
m = "{0:10.4f}".format(y)
print(m)

# { 또는 } 문자 표현하기
n = "{{ and }}".format()
print(n)
```

## 문자열 관련 함수

### 문자 개수 세기(count)

```python
a = "hi, my name is kwangwoojae"
print(a.count('a'))
```

### 위치 알려주기1(find)

```python
a = "hi, my name is kwangwoojae"
print(a.find('n'))
print(a.find('x'))
```

### 위치 알려주기2(index)

```python
a = "hi, my name is kwangwoojae"
print(a.index('y')) 
# print(a.index('x')) 
```

### 문자열 삽입(join)

```python
print("abcd")
print(",".join("abcd"))
print(['a', 'b', 'c', 'd'])
print(",".join(['a', 'b', 'c', 'd']))
```

### 소문자를 대문자로 바꾸기(upper)

```python
a = 'hi'
print(a.upper())
```

### 대문자를 소문자로 바꾸기(lower)

```python
a = 'HI'
print(a.lower())
```

### 왼쪽 공백 지우기(lstrip)

```python
a = f"{'hi':^10}"
print(a)
print(a.lstrip())
```

### 오른쪽 공백 지우기(rstrip)

```python
a = f"{'hi':<10}"
print(a + "woojae")
print(a.rstrip() + "woojae")
```

### 양쪽 공백 지우기(strip)

```python
a = f"{'hi':^10}"
print("!" + a + "!")
print("!" + a.strip() + "!")
```

### 문자열 바꾸기(replace)

```python
a = "Life is too short"
print(a)
print(a.replace("Life", "My leg"))
```

### 문자열 나누기(split)

```python
a = "Life is too short"
print(a)
print(a.split())
b = "a:b:c:d"
print(b)
print(b.split(':'))
```

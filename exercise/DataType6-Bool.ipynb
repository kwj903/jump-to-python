# 불 자료형 학습과 연습

## 불 자료형이란?

```python
a = True
b = False

print(type(a))
print(type(b))

print(1 == 1)
print(1 < 2) 
print(1 > 2)
```

## 자료형의 참과 거짓

```python
a = [1, 2, 3, 4]
while a:
  print(a.pop())
  
if []:
  print("참")
else:
  print("거짓")

if[1, 2, 3, 4]:
  print("참")
else:
  print("거짓")  
```

## 불 연산

```python
print(bool('python'))
print(bool())

print(bool([1, 2, 3, 4]))
print(bool([]))
print(bool(0))
print(bool(1))
```

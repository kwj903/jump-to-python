# Q9

```python
# 답은 3번 입니다. 왜냐하면 딕셔너리의 key는 변경 불가능한 자료형이어야 하는데 리스트는 변경 가능한 자료형이기 때문입니다. 따라서 리스트를 key로 사용할 수 없습니다.
a = dict()
print(a)
a['name'] = 'python' # 1번
print(a)
a[("a",)] = "python"  # 2번
print(a)
# a[[1]] = 'python' # 3번
# print(a)
a[250] = "python"  # 4번
print(a)
```

## 함수를사용하는 이유는 무엇일까?

1. 반복적으로 사용되는 가치 있는 부분을 한 뭉치로 묶어서 어떤 입력값을 주었을 때 어떤 결괏값을 돌려준다라는 식의 함수로 작성하기 위해 함수가 필요.
2. 자신이 만든 프로그램을 함수화하면 프로그램 흐름을 일목요연하게 볼 수 있기 때문.

## 매개변수와 인수

## 입력값과 결괏값에 따른 함수의 형태

입력값 -> 함수 -> 결과값  
함수의 형태는 입력값과 결과값의 존재 유무에 따라 4가지 유형으로 나뉨

### 일반적인 함수

````python
def add(a, b):
    result = a + b
    return result   # a+b의 결괏값 반환

a = add(3, 4)
print(a)

# 입력값과 결괏값이 있는 함수의 사용법
# 결과값을 받을 함수 = 함수이름(입력변수 1, 입력변수 2, ...)
````

### 입력값이 없는 함수

````python
def say():
    return 'Hi'

a = say()
print(a)

# 입력값이 없고 결괏값만 있는 함수의 사용법
# 결과값을 받을 변수 = 함수이름()
````

### 결괏값이 없는 함수

````python
def add(a, b):
    print("%d, %d의 합은 %d입니다." % (a, b, a+b))
    
add(3, 4)

a = add(3, 4)
print(a)

# 입력값이 있고 결괏값이 없는 함수의 사용법
# 함수이름(입력변수 1, 입력변수 2, ...)
````

### 입력값도 결괏값도 없는 함수

````python
def say():
    print('Hi')

say()

# 입력값이 없고 결괏값이 없는 함수의 사용법
# 함수이름()
````

## 매개변수 지정하여 호출하기

## 입력값이 몇 개가 될지 모를 때는 어떻게 해야 할까?

```python
def 함수이름(*매개변수):
    수행할 문장
```
 이런 형태로 만들어주면 된다.

### 여러 개의 입력값을 받는 함수 만들기

````python
def add_many(*args):
    result = 0
    for i in args:
        result = result + i
    return result

result = add_many(1, 2, 3)
print(result)
````

````python
def add_mul(choice, *args):
    if choice == "add":
        result = 0
        for i in args:
            result = result + i
    elif choice == "mul":
        result = 1
        for i in args:
            result = result * i
    return result

result = add_mul('add', 1, 2, 3, 4, 5)
print(result)

result = add_mul('mul', 1, 2, 3, 4, 5)
print(result)
````

- ***점프 투 파이썬*** : 키워드 파라미터

````python
def print_kwargs(**kwargs):
    print(kwargs)

print_kwargs(a=1)
print_kwargs(name='foo', age=3)

a = print_kwargs(a=1)
print(a)
a = print_kwargs(name="foo", age=3)
print(a)

# 매개변수 이름 앞에 *이 붙으면 튜플로, **이 붙으면 딕셔너리로 입력값을 받는다.
````

## 함수의 결괏값은 언제나 하나이다

````python
def add_and_mul(a, b):
    return a+b, a*b
result = add_and_mul(3, 4)
print(result)

# 이처럼 리턴값이 여러 개인 경우 튜플로 묶어서 반환한다.
````

````python
def add_and_mul(a, b):
    return a+b
    return a*b

result = add_and_mul(3, 4)
print(result)
````

````python
def add_and_mul(a, b):
    return a+b
result = add_and_mul(3, 4)
print(result)

# return문을 만나면 함수를 빠져나가므로, 위의 예에서 a*b는 실행되지 않는다. 
````

- ***점프 투 파이썬*** : return의 또다른 쓰임새

````python
def say_nick(nick):
    if nick == '바보':
        return
    print('나의 별명은 %s입니다.' % nick)
    
say_nick("야호")
say_nick("바보")

# return문을 사용해서 함수를 빠져나가는 방법은 실제 프로그래밍에서 자주 사용한다.
````

## 매개변수에 초깃값 미리 설정하기

````python
def say_myself(name, old, man=True):
    print("나의 이름은 %s입니다." % name)
    print("나이는 %d입니다." % old)
    if man:
        print("남자입니다.")
    else:
        print("여자입니다.")
        
say_myself("곽우재", 34)
say_myself("곽우재", 34, True)
say_myself("곽우재", 34, False)
````

## 함수 안에서 선언한 변수의 효력 범위

````python
a = 1
def vertest(a):
    a = a + 1
    return a
print(vertest(a))
print(a)

# 위의 함수와 동일한 함수이다.
def vertest(hello):
    hello = hello + 1
    
````

````python
a = 100
def vartest(a):
    a = a + 1
print(vartest(3))
print(a)

# 함수 안에서 선언한 매개변수는 함수 안에서만 사용할 수 있는 지역변수이다.
````

### 함수 안에서 함수 밖의 변수를 변경하는 방법

#### 1. return 사용하기

````python
a = 1
def vartest(a):
    a = a + 1
    return a

a = vartest(a)
print(a)
````

#### 2. global 명령어 사용하기

````python
a = 1
def vartest():
    global a
    a = a + 1
vartest()
print(a)    
````

## lambda

````python
# def를 사용할 정도로 복잡하지 않거나 def를 사용할 수 없는 상황에서는 lambda를 사용해서 함수를 간단히 만들 수 있다.
# lambda 매개변수1, 매개변수2, ... : 매개변수를 이용한 표현식
add = lambda a, b: a + b
result = add(3, 4)
print(result)
````

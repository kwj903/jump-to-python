## 패키지란 무엇인가?

패키지(Packages)는 도트(.)를 사용하여 파이썬 모듈을 계층적(디렉터리 구조)으로 관리할 수 있게 해준다.

## 패키지 만들기

3. echo.py 파일은 다음과 같이 만든다.

## ```__init__.py```의 용도

 ```__init__.py```의 용도는 해당 디렉터리가 패키지의 일부임을 알려주는 역할을 한다. 구버전 파이썬에는 이 파일이 패키지인식을 위한 파일이었지만 패치되고서는  
 하위버전 호환을 위해 만들어주는 것이 좋다.

````python
# echo를 임폴트 하던 흔적 찾기 (실습을 위해 이전에 캐시등으로 남아서 자동으로 연결되는 현상을 막기 위해 하는 확인과정이다.) 
print("echo in globals?", "echo" in globals())
if "echo" in globals():
    import inspect

    print("echo object:", echo)
    print("echo file:", getattr(echo, "__file__", None))
````

````python
import sys

# 1) globals에 남아있는 이름 제거
globals().pop("echo", None)

# 2) 패키지 객체에 붙어있는 속성 제거 (붙어있을 수 있음)
import game.sound

if hasattr(game.sound, "echo"):
    delattr(game.sound, "echo")

# 3) sys.modules 캐시 제거
for k in ["game.sound.echo", "game.sound"]:
    sys.modules.pop(k, None)
````

## relative 패키지

````python
# render.py
from game.sound.echo import echo_test
def render_test():
    print("render")
    echo_test()
````

````python
from game.graphic.render import render_test
render_test()
````

````python
# 다음과 같이 relative하게 import하는 것도 가능하다. 왜냐하면 아래 코드가 입력되는 모듈과 해당 모듈에서 임포트시킬 echo모듈은 같은 부모 디렉토리를 갖고 있기 때문에 ..(부모디렉토리를 뜻함)과 같은 relative한 접근자를 사용할 수 있다.
# 이러한 사용법은 같은 부모 디렉토리를 가진 모듈일 경우 가능한 방법이다.
# 현재 이코드는 해당 디렉토리에 있는 모듈이 아님으로 당연히 오류가 난다. 하지만 아래 셀에서 결과를 실행시켜보면 가능 한것을 알 수 있다.
# render.py
from ..sound.echo import echo_test


def render_test():
    print("render")
    echo_test()
````

## 클래스는 왜 필요한가?

- 클래스가 필요한 이유는 반복요구되면서 다르게 관리되야하는 경우 함수를 중복적으로 만들어야하는데 클래스를 이용하면 하나의 클래스를 만들어 호출 후  
  인스턴스화해서 각각의 객체를 만들어 사용하면 중복적으로 만들필요 없어지기 때문이다.

````python
result = 0
def add(num):
    global result
    result += num
    return result

print(add(3))
print(add(4))
````

````python
result1 = 0
result2 = 0


def add1(num):
    global result1
    result1 += num
    return result1


def add2(num):
    global result2
    result2 += num
    return result2



print(add1(3))
print(add1(4))
print(add2(3))
print(add2(7))
````

````python
class Calculator:
    def __init__(self):
        self.result = 0
        
    def add(self, num):
        self.result += num
        return self.result
    
    def sub(self, num):
        self.result -= num
        return self.result
    
cal1 = Calculator()
cal2 = Calculator()

print(cal1.add(3))
print(cal1.add(4))
print(cal2.add(3))
print(cal2.add(7))
````

### 곱하기, 빼기, 나누기 기능 만들기

````python
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        result = self.first + self.second
        return result
    
    def mul(self):
        result = self.first * self.second
        return result
    
    def sub(self):
        result = self.first - self.second
        return result
    
    def div(self):
        result = self.first / self.second
        return result
    
a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 8)
print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
print(b.add())
print(b.mul())
print(b.sub())
print(b.div())
````

## 생성자(Constructor)

````python
a = FourCal()
a.add()
````

````python
class FourCal:
    def __init__(self, first, second):
        self.first = first
        self.second = second

    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        result = self.first + self.second
        return result

    def mul(self):
        result = self.first * self.second
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def div(self):
        result = self.first / self.second
        return result
````

````python
a = FourCal()
````

````python
a = FourCal(4, 2)
print(a.__dict__)
print(a.first)
print(a.second)
print(a.add())
print(a.div())
````

## 클래스의 상속

````python
class MoreFourCal(FourCal):
    pass

a = MoreFourCal(8, 4)
print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
````

- ***점프 투 파이썬*** : 왜 상속을 해야 할까?

 보통 상속은 기존 클래스에 기능을 추가하거나 기능을 변경하려고 할때 사용한다.기존 클래스가 라이브러리 형태로 제공 되거나 수정이 허용되지 않는 경우가 있기 대문이다.

````python
class MoreFourCal(FourCal):
    def pow(self):
        result = self.first ** self.second
        return result
    
a = MoreFourCal(4, 2)
print(a.pow())
````

### 메서드 오버라이딩

````python
a = FourCal(4, 0)
a.div()
````

````python
class SafeFourCal(FourCal):
    def div(self):
        if self.second == 0:
            return 0
        else:
            return self.first / self.second
````

````python
a = SafeFourCal(4, 0)
print(a.div())
````

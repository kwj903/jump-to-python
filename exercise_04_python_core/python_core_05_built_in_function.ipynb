## ***filter***

```python
#positive.py
def positive(l):
    result = []
    for i in l:
        if i > 0:
            result.append(i)
    return result

print(list(positive([1, -3, 2, 0, -5, 6])))
        
```

```python
# filter1.py
def positive2(x):
    return x > 0
print(list(filter(positive2, [1, -3, 2, 0, -5, 6])))
```

```python
print(list(filter(lambda x: x > 0, [1, -3, 2, 0, -5, 6])))
```

## ***map***

```python
# two_times.py
def two_times(numberList):
    result = []
    for number in numberList:
        result.append(number*2)
    return result

result = two_times([1, 2, 3, 4])
print(result)
```

```python
def two_times2(x): return x * 2
list(map(two_times2, [1, 2, 3, 4]))
```

```python
list(map(lambda x: x*2, [1, 2, 3, 4]))
```

## ***sorted***

```python
# sort는 정렬된 값을 반환하지 않는다. 다만 이미 있는 리스트 자료형을 정렬할 뿐이다.
print([3, 4, 1, 6].sort())
a = [3, 4, 1, 6].sort()
print(a)
a = [3, 4, 1, 6]
print(a)
a.sort()
print(a)
```

```python
# 반면 sorted는 정렬 후 그 결과를 리스트로 돌려준다
print(sorted([3, 4, 1, 6, 2]))
b = sorted([3, 4, 1, 6, 2])
print(b)
print(sorted(["a", "b", "c"]))
print(sorted("zero"))
print(sorted((3, 2, 1)))
```

## Q1 다음은 Calculator 클래스이다.

```python
class Calulator:
    def __init__(self):
        self.value = 0
    
    def add(self, val):
        self.value += val
    self.value += 3
```

## 위 클래스를 상속하는 UpgradeCalculator를 만들고 값을 뺄 수 있는 minus 메서드를 추가해 보자.  
## 즉 다음과 같이 동작하는 클래스를 만들어야 한다.

```python
cal = UpgradeCalculator()
cal.add(10)
cal.minus(7)

print(cal.value)
```

```python
# 문제 답
class Upgradecalculator(Calulator):
    def minus(self, val):
        self.value -= val

cal = Upgradecalculator()
cal.add(10)
cal.minus(7)

print(cal.value)
```

## Q2 객체 변수 value가 100 이상의 값은 가질 수 없도록 제한하는 MaxLimitCalculator 클래스  
## 를 만들어 보자. 즉 다음과 같이 동작해야 한다.

```python
cal = MaxLimitCalculator()
cal.add(50)
cal.add(60)

print(cal.value)
```

## 단 반드시 다음과 같은 Calculator 클래스를 상속해서 만들어야한다.

```python
class Calculator:
    def __init__(self):
        self.value = 0
        
    def add(self, val):
        self.value += val
```

```python
# 문제 답
class MaxLimitCalculator(Calculator):
    def add(self, val):
        self.value += val
        if self.value > 100:
            self.value = 100
        
cal = MaxLimitCalculator()
cal.add(50)
cal.add(60)

print(cal.value)
```

## Q13 random 모듈을 사용하여 로또 번호(1~45 사이의 숫자 6개)를 생성해 보자(단 중복된 숫자가 있으면 안 됨.)

```python
import random
result = []
numbers = [i for i in range(1, 46)]
def rotto(numbers):
    for i in range(6):
        number = random.choice(numbers)
        numbers.remove(number)
        # print(numbers)
        result.append(number)
    return result
    
rotto_num = rotto(numbers)
print("행운은 번호는?: ", *rotto_num)
        
```

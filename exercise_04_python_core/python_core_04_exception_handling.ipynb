## 오류 예외 처리 기법

### try, except문

````python
# 아래는 오류 처리를 위한 try, except문의 기본 구문이다.
try:
    ...
except [발생 오류[as 오류 메시지 변수]]:
    ...

# 아래 구문을 보면 [] 기호를 사용하는데, 이 기호는 괄호 안의 내용을 생략할 수있따는 관례 표기법이다.    
# except [발생 오류[as 오류 메시지 변수]]:
````

1. try, except만 쓰는 방법

````python
# 이 경우는 오류 종류에 상관없이 오류가 발생하면 except 블록을 수행한다.
try:
    ...
except:
    ...
````

2. 발생 오류만 포함한 except문

````python
# 이 경우는 오류가 발생했을 때 except문에 미리 정해 놓은 오류 이름과 일치 할 때만 except블록을 수행한다는 뜻이다.
try:
    ...
except 발생 오류:
````

3. 발생 오류와 오류 메시지 변수까지 포함한 except문

````python
# 이 경우는 두 번째 경우에서 오류 메시지의 내용까지 알고 싶을 때 사용하는 방법이다.
try:
    ...
except 발생 오류 as 오류 메시지 변수:
    ...
````

````python
try:
    4 / 0
except ZeroDivisionError as e:
    print(e)
````

- ***나 혼자 코딩*** : 앞에서 나온 3번 방법을 사용해 IndexError가 발생할 때 오류 메시지를 출력하는 소스를 작성해 보자.

````python
try:
    a = [1, 2, 3, 4]
    print(a[4])
except IndexError as e:
    print(e)
````

### try ... finally

````python
try:
    f = open('foo.txt', 'r')
    # 무엇인가를 수행
    a = f.read()
except FileNotFoundError as e:
    print(e)
finally:
    f.close()
````

### 여러 개의 오류 처리하기

````python
try:
    ...
except 발생 오류 1:
    ...
except 발생 오류 2:
    ...
````

````python
# 예시

try:
    4 / 0
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
except IndexError:
    print("인덱싱할 수 없습니다.")
    
````

````python
# 예시

try:
    a =[1,2]
    print(a[3])
    4/0
except ZeroDivisionError as e:
    print(e)
except IndexError as e:
    print(e)
    
````

````python
# 예시

try:
    a = [1, 2]
    print(a[3])
    4/0
except (ZeroDivisionError, IndexError) as e:
    print(e)
````

## 오류 일부러 발생시키기

- 파이썬 ```raise```명령어를 사용하여 반드시 정의해야 하는 함수에 오류를 발생시키도록 할 수 있다.

````python
class Bird:
    def fly(self):
        raise NotImplementedError
    
class Eagle(Bird):
    pass

eagle = Eagle()
eagle.fly()
````

````python
class Eagle(Bird):
    def fly(self):
        print("very fast")
        
eagle = Eagle()
eagle.fly()
````

## 예외 만들기

````python
class MyError(Exception):
    pass
````

````python
def say_nick(nick):
    if nick == '바보':
        raise MyError()
    print(nick)
````

````python
say_nick("천사")
say_nick("바보")
````

````python
try:
    say_nick("천사")
    say_nick("바보")
except MyError:
    print("허용되지 않는 별명입니다.")
````

````python
try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
````

````python
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다."
````

````python
try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
````

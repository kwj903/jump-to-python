- ***점프 투 파이썬*** : __name__ 변수란?

 파이썬의 __name__ 변수는 파이썬이 내부적으로 사용하는 특수한 변수이다. 어떤 모듈을 python 명령어로 실행시키면 __name__변수에 자동으로 __main__값이 저장된다.  
 하지만 모듈을 import를 할 경우에는 __name__변수에는 해당모듈의 확장자가 빠진 이름 값이 저장된다.

## 클래스나 변수 등을 포함한 모듈

```python
# mod2.py
PI = 3.141592

class Math:
    def solv(self, r):
        return PI * (r ** 2)
    
def add(a, b):
    return a + b
```

```python
%cd /Users/kwj903/Downloads/workspace/education/my_educate/jump-to-python/exercise_04_python_core/module_exercise_dir
# %cd /Users/kwj903/Downloads/workspace/education/my_educate/jump-to-python/exercise_04_python_core/
import mod2
print(mod2.PI)
!python3 -c "import mod2; print(mod2.PI)"
```

```python
a = mod2.Math()
print(a.solv(2))
!python3 -c "import mod2; a = mod2.Math(); print(a.solv(2))"
```

```python
print(mod2.add(mod2.PI, 4.4))
!python3 -c "import mod2; print(mod2.add(mod2.PI, 4.4))"
```
